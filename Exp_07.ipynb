```python
print(df.dropna(how="all", axis=0).shape)   # drop a row only if ALL its values are null
print()
print(df.dropna(how="any", axis=0).shape)   # drop a row if it contains ANY 1 null value

print(df.dropna(how="all", axis=1).shape)   # drop a column only if ALL its values are null
print()
print(df.dropna(how="any", axis=1).shape)   # drop a column if it contains ANY 1 null value
```

```python
df.dropna(thresh=11).shape
# Count-based: a row needs at least 11 non-null values (out of 12 columns) to be kept
# Rows with fewer valid values than thresh get dropped

# Filling missing values
print("Missing values: ")
print(df.isnull().sum())

df_copy = df.copy()   # work on a copy so the original is unchanged

# Embarked (text): fill with the mode (most frequent value)
df_copy["Embarked"] = df_copy["Embarked"].fillna(df_copy["Embarked"].mode()[0])

# Age: fill with the mean
df_copy["Age"] = df_copy["Age"].fillna(df_copy["Age"].mean())

# Age median version (saved in a separate column to compare with the mean)
df_copy["Age_median"] = df["Age"].fillna(df["Age"].median())

# Cabin: too many missing, so fill with a constant label
df_copy["Cabin"] = df_copy["Cabin"].fillna("Unknown")

print(df_copy.isnull().sum())
df_copy.head()
```

```python
import random

# 1. Get a clean list of all valid Age values
valid_ages = df["Age"].dropna().tolist()

# 2. Replace missing values with a random choice, otherwise keep the value
df["Age"] = [
    random.choice(valid_ages) if pd.isna(x) else x for x in df["Age"]
]
df["Age"].isnull().sum()   # should now be 0
```

```python
# 1. Identify Outliers using IQR (on the real Fare column)
df = titanic.copy()

Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
print(df["Fare"].describe())
IQR = Q3 - Q1

# Define IQR boundaries
lower_iqr = Q1 - 1.5 * IQR
upper_iqr = Q3 + 1.5 * IQR

# Create a mask for rows that IQR flags as outliers
outlier_mask = (df["Fare"] < lower_iqr) | (df["Fare"] > upper_iqr)
print("Number of outliers:", outlier_mask.sum())
```

```python
# 2. Calculate Z-score boundaries based ONLY on the clean, valid data
clean_data = df.loc[~outlier_mask, "Fare"]
mean = clean_data.mean()
std = clean_data.std()

# Define Z-score limits (3 standard deviations)
lower_z_limit = mean - (3 * std)
upper_z_limit = mean + (3 * std)
print(lower_z_limit, upper_z_limit)
```

```python
# 3. Impute: cap the IQR-flagged outliers at the Z-score boundaries
df["Fare"] = np.where(
    df["Fare"] > upper_iqr,
    upper_z_limit,
    np.where(df["Fare"] < lower_iqr, lower_z_limit, df["Fare"]),
)
print("Max Fare before:", titanic["Fare"].max())
print("Max Fare after :", df["Fare"].max())
df["Fare"]
```

# Standardisation and Normalisation

```python
import time


# Age has missing values, so fill them first (median) to avoid NaN in the results
df["Age"] = df["Age"].fillna(df["Age"].median())
df.head()
```

```python
# columns to be processed
features = ["Age", "Fare", "SibSp", "Parch"]

# --- PROCESS 1: STANDARDISATION TIMER ---
start_std = time.perf_counter()
for col in features:
    mean_val = df[col].mean()
    std_val = df[col].std()
    # z-score: mean becomes 0 and std becomes 1
    df[f"{col}_standardised"] = (df[col] - mean_val) / std_val
end_std = time.perf_counter()
std_time_ms = (end_std - start_std) * 1000   # seconds -> milliseconds
```

```python
# --- PROCESS 2: NORMALISATION TIMER ---
start_norm = time.perf_counter()
for col in features:
    min_val = df[col].min()
    max_val = df[col].max()
    # Min-Max normalisation: scales values between 0 and 1
    df[f"{col}_normalised"] = (df[col] - min_val) / (max_val - min_val)
end_norm = time.perf_counter()
norm_time_ms = (end_norm - start_norm) * 1000
```

```python
# Print benchmark results
print(f"Standardisation Time : {std_time_ms:.4f} ms")
print(f"Normalisation Time   : {norm_time_ms:.4f} ms")
print(f"Total Pipeline Time  : {std_time_ms + norm_time_ms:.4f} ms")

df.head()
```

# Merging

```python
# df1: first 5 passengers (PassengerId 1-5) with personal details
df1 = titanic[["PassengerId", "Name", "Sex", "Age"]].iloc[0:5].copy()

# df2: PassengerId 3-7 with ticket details (only Ids 3, 4, 5 overlap with df1)
df2 = titanic[["PassengerId", "Pclass", "Fare", "Embarked"]].iloc[2:7].copy()

print("--- Original DataFrames ---")
print("Passenger DataFrame:\n", df1)
print("Ticket DataFrame:\n", df2)
```

```python
# INNER MERGE: Keeps only rows where PassengerId exists in BOTH DataFrames.
inner_merge = pd.merge(df1, df2, on='PassengerId', how='inner')
print("--- Inner Merge ---")
print(inner_merge, "\n")

# OUTER MERGE: Keeps ALL rows from both DataFrames, filling missing values with NaN.
outer_merge = pd.merge(df1, df2, on='PassengerId', how='outer')
print("--- Outer Merge ---")
print(outer_merge, "\n")

# LEFT MERGE: Keeps all rows from the left (passenger) DataFrame, matching right where possible.
left_merge = pd.merge(df1, df2, on='PassengerId', how='left')
print("--- Left Merge ---")
print(left_merge, "\n")

# RIGHT MERGE: Keeps all rows from the right (ticket) DataFrame, matching left where possible.
right_merge = pd.merge(df1, df2, on='PassengerId', how='right')
print("--- Right Merge ---")
print(right_merge)
```
