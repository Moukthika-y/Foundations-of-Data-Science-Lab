```python
students = pd.DataFrame({
    'Student_ID': [101, 102, 103, 104],
    'Name': ['Asha', 'Rahul', 'Priya', 'Kiran'],
    'Department': ['CSE', 'CSE', 'ECE', 'ECE']
})

marks = pd.DataFrame({
    'Student_ID': [101, 102, 103, 105],
    'Marks': [85, 78, 92, 75]
})
```

```python
result = pd.merge(students, marks, on='Student_ID', how='inner')

print(result)
```

```python
result = pd.merge(students, marks, on='Student_ID', how='left')

print(result)
```

```python
result = pd.merge(students, marks, on='Student_ID', how='right')

print(result)
```

```python
result = pd.merge(students, marks, on='Student_ID', how='outer')

print(result)
```

```python
students = pd.DataFrame({
    'ID': [101, 102, 103],
    'Name': ['Asha', 'Rahul', 'Priya']
})

marks = pd.DataFrame({
    'Student_ID': [101, 102, 103],
    'Marks': [85, 78, 92]
})
```

```python
result = pd.merge(
    students,
    marks,
    left_on='ID',
    right_on='Student_ID'
)

print(result)
```

```python
df = pd.DataFrame({
    'Student': ['Asha', 'Rahul', 'Priya'],
    'Maths': [85, 78, 92],
    'Python': [90, 82, 95],
    'DBMS': [88, 80, 91]
})

print(df)
```

```python
long_df = pd.melt(
    df,
    id_vars=['Student'],
    var_name='Subject',
    value_name='Marks'
)

print(long_df)
```

```python
pivot_df = long_df.pivot(
    index='Student',
    columns='Subject',
    values='Marks'
)

print(pivot_df)
```
